# Cek folder Oceanic Life sebelum training
Jalankan sel berikut di VS Code setelah notebook 01. Notebook ini **tidak menjalankan CNN**, hanya memeriksa gambar dan status persetujuan label.


```python
from pathlib import Path
import csv
from collections import Counter

p=Path.cwd().resolve()
if not (p/'train.py').is_file():
    for q in [p, *p.parents, Path.home()/'Downloads'/'Oceanic_Life_CNN_PERBAIKAN_FOLDER',
              Path.home()/'Downloads'/'Oceanic_Life_CNN_Tugas_Lengkap']:
        if (q/'train.py').is_file(): p=q; break
root=p/'hasil_pengelompokan'
classes=['Fish','Coral','Shark','Octopuses','Crustaceans','Seaweed','Sponges','Other marine organisms']
exts={'.jpg','.jpeg','.png','.webp','.bmp','.tif','.tiff'}
print('Folder proyek:',p)
for name,loc in [('FINAL',root/'Oceanic_Life-Dataset'),('USULAN_AI',root/'Usulan_AI'/'Oceanic_Life-Dataset'),
                 ('SUMBER_DOWNLOADS',Path.home()/'Downloads'/'Oceanic_Life-Dataset')]:
    print('\n',name,'→',loc,'ADA:',loc.is_dir())
    print({cls:sum(f.is_file() and f.suffix.lower() in exts for f in (loc/cls).rglob('*'))
           if (loc/cls).is_dir() else 0 for cls in classes})
    if loc.is_dir():
        others=[f for f in loc.rglob('*') if f.is_file() and f.suffix.lower() in exts]
        print('Total gambar pada semua subfolder:',len(others))
        for f in others[:5]: print(' -',f.relative_to(loc))
review=root/'review_labels.csv'
print('\nFile pemeriksaan label:',review,'ADA:',review.is_file())
if review.is_file():
    with review.open(encoding='utf-8-sig',newline='') as fh:
        rows=list(csv.DictReader(fh))
    print('Jumlah baris CSV:',len(rows))
    print('Status:',dict(Counter(r.get('status','') for r in rows)))
    print('Label final disetujui:',dict(Counter(r.get('kelas_final','') for r in rows
       if r.get('status','').strip().upper() in {'DISETUJUI_MANUAL','DISETUJUI_DARI_LABEL_ASLI'})))
print('\nJika FINAL=0 dan USULAN_AI>0, periksa review_labels.csv sebelum Notebook 02.')
```
